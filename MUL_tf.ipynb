```python
import torch
import subprocess
import sys

print("="*50)
print("CUDA VERSION DETECTION")
print("="*50)

# Method 1: Using PyTorch (most reliable if PyTorch is installed)
if torch.cuda.is_available():
    print(f"PyTorch sees GPU: {torch.cuda.get_device_name(0)}")
    print(f"PyTorch CUDA version: {torch.version.cuda}")
    print(f"PyTorch cuDNN version: {torch.backends.cudnn.version()}")
else:
    print("PyTorch does not detect a GPU.")

# Method 2: Using nvidia-smi command (works on all systems with NVIDIA driver)
try:
    result = subprocess.run(['nvidia-smi', '--query-gpu=name,driver_version,compute_cap', '--format=csv,noheader'], 
                            capture_output=True, text=True, check=True)
    print("\nnvidia-smi output:")
    print(result.stdout.strip())
except Exception as e:
    print(f"Could not run nvidia-smi: {e}")

# Method 3: Check environment variables (if CUDA toolkit is installed)
import os
cuda_path = os.environ.get('CUDA_PATH', os.environ.get('CUDA_HOME', 'Not set'))
print(f"\nCUDA_PATH environment variable: {cuda_path}")

# Optional: Try to find cuda version from toolkit
if cuda_path != 'Not set':
    version_file = os.path.join(cuda_path, 'version.txt')
    if os.path.exists(version_file):
        with open(version_file, 'r') as f:
            print(f"CUDA toolkit version from file: {f.read().strip()}")

# Method 4: Using TensorFlow (if installed)
try:
    import tensorflow as tf
    print(f"\nTensorFlow version: {tf.__version__}")
    print(f"TensorFlow built with CUDA: {tf.test.is_built_with_cuda()}")
    if tf.test.is_built_with_cuda():
        # TensorFlow doesn't directly expose CUDA version, but we can get device info
        gpus = tf.config.list_physical_devices('GPU')
        if gpus:
            print(f"TensorFlow sees GPUs: {[gpu.name for gpu in gpus]}")
        else:
            print("TensorFlow does not detect any GPU.")
except ImportError:
    print("\nTensorFlow not installed.")

print("\n" + "="*50)
```
